Part B- Read, Count & Search

```python
# Read and print numbered lines
with open("python_notes.txt", "r", encoding="utf-8") as f:
  file_lines = f.readlines()

  print("---Content of python_nots.txt---")
  for i, line in enumerate(file_lines, start=1):
    print(f"{i}. {line.rstrip()}")

    print(f"\nTotal lines: {len(file_lines)}")
```

```python
# Keyword search (case insensitive)

keyword = input("Enter a keyword to search:").strip()
matches = [line.rstrip()for line in file_lines if keyword.lower() in line.lower()]

if matches:
  print(f"\nLines containing '{keyword}':")
  for match in matches:
    print(" ",match)
  else:
      print(f"No lines found containing '{keyword}'.")
```

```python
import requests
BASE_URL = "https://dummyjson.com/products"
```

Step 1- Fetch & Display 20 Products

```python
products = []

try:
    response = requests.get(f"{BASE_URL}?limit=20", timeout=5)
    data = response.json()
    products = data["products"]

    print(f"{'ID':<4} | {'Title':<32} | {'Category':<14} | {'Price':<9} | {'Rating'}")
    print("-" * 4 + "-|-" + "-" * 32 + "-|-" + "-" * 14 + "-|-" + "-" * 9 + "-|-" + "-" * 6)
    for p in products:
        print(f"{p['id']:<4} | {p['title'][:32]:<32} | {p['category'][:14]:<14} | ${p['price']:<8.2f} | {p['rating']}")

except requests.exceptions.ConnectionError:
    print("Connection failed. Please check your internet.")
except requests.exceptions.Timeout:
    print("Request timed out. Try again later.")
except Exception as e:
    print(f"Unexpected error: {e}")
```

Step 2- Filter (Rating >=4.5) & Sort by Price Descending

```python
filtered = [p for p in products if p["rating"] >= 4.5]
filtered.sort(key=lambda p: p["price"], reverse=True)

print("Products with rating >= 4.5, sorted by price (high to low):\n")
print(f"{'ID':<4} | {'Title':<32} | {'Price':<9} | {'Rating'}")
print("-" * 60)
for p in filtered:
    print(f"{p['id']:<4} | {p['title'][:32]:<32} | ${p['price']:<8.2f} | {p['rating']}")
```

Step 3- Search by Category: Laptops

```python
try:
    resp = requests.get(f"{BASE_URL}/category/laptops", timeout=5)
    laptops = resp.json()["products"]

    print(f"Laptops ({len(laptops)} found):\n")
    for p in laptops:
        print(f"  {p['title']:<40} ${p['price']:.2f}")

except requests.exceptions.ConnectionError:
    print("Connection failed. Please check your internet.")
except requests.exceptions.Timeout:
    print("Request timed out. Try again later.")
except Exception as e:
    print(f"Unexpected error: {e}")
```

Step 4- POST Request (Simulated Add Product)

```python
new_product = {
    "title": "My Custom Product",
    "price": 999,
    "category": "electronics",
    "description": "A product I created via API"
}

try:
    resp = requests.post(f"{BASE_URL}/add", json=new_product, timeout=5)
    print("POST Response:")
    print(resp.json())

except requests.exceptions.ConnectionError:
    print("Connection failed. Please check your internet.")
except requests.exceptions.Timeout:
    print("Request timed out. Try again later.")
except Exception as e:
    print(f"Unexpected error: {e}")
```

Task 3- Exception Handling

Part A- Guarded Calculator

```python
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return "Error: Cannot divide by zero"
    except TypeError:
        return "Error: Invalid input types"

print(safe_divide(10, 2))       # 5.0
print(safe_divide(10, 0))       # ZeroDivisionError
print(safe_divide("ten", 2))    # TypeError
```

Part B- Guarded File Reader

```python
def read_file_safe(filename):
  try:
      with open(filename, "r", encoding="utf-8")as f:
        content = f.read()
      return content
  except FileNotFoundError:
        print(f"Error: File '{filename}' not found.")
        return None
  finally:
          print ("File operation attempt complete.")

print("---Reading python_notes.txt---")
content = read_file_safe("python_notes.txt")
if content:
    print(content[:200], "...")

print("\n---Reading ghost_file.txt---")
read_file_safe("ghot_file.txt")
```

Part C- Robust API Calls (fetch_product helper)

```python
def fetch_product(product_id):
    try:
        resp = requests.get(f"{BASE_URL}/{product_id}", timeout=5)
        if resp.status_code == 404:
            print(f"Product {product_id} not found (HTTP 404).")
            return None
        return resp.json()
    except requests.exceptions.ConnectionError:
        print("Connection failed. Please check your internet.")
    except requests.exceptions.Timeout:
        print("Request timed out. Try again later.")
    except Exception as e:
        print(f"Unexpected error: {e}")

p = fetch_product(1)
if p:
    print(f"Fetched: {p['title']} — ${p['price']}")
```

Part D- Input Validation Loop

```python
while True:
    user_input = input("Enter a product ID to look up (1-100), or 'quit' to exit: ").strip()

    if user_input.lower() == "quit":
        print("Exiting lookup.")
        break

    try:
        product_id = int(user_input)
    except ValueError:
        print("  Warning: Invalid input — please enter a whole number between 1 and 100.")
        continue

    if not (1 <= product_id <= 100):
        print("  Warning: Out of range — please enter a number between 1 and 100.")
        continue

    try:
        resp = requests.get(f"{BASE_URL}/{product_id}", timeout=5)
        if resp.status_code == 404:
            print("  Product not found.")
        elif resp.status_code == 200:
            p = resp.json()
            print(f"  {p['title']} — ${p['price']}")
        else:
            print(f"  Unexpected status code: {resp.status_code}")
    except requests.exceptions.ConnectionError:
        print("  Connection failed. Please check your internet.")
    except requests.exceptions.Timeout:
        print("  Request timed out. Try again later.")
    except Exception as e:
        print(f"  Unexpected error: {e}")
```

Task 4- Logging to File

```python
from datetime import datetime
```

```python
def log_error(message):
    with open("error_log.txt", "a") as f:
        time = datetime.now()
        f.write(f"[{time}] ERROR: {message}\n")

# trigger connection error
try:
    requests.get("https://this-host-does-not-exist-xyz.com/api")
except Exception:
    log_error("ConnectionError — No connection")

# trigger 404 manually
response = requests.get("https://dummyjson.com/products/999")

if response.status_code != 200:
    log_error("HTTPError — 404 Not Found for product ID 999")

# print log file
with open("error_log.txt", "r") as f:
    print("\nError Log:\n")
    print(f.read())
```
